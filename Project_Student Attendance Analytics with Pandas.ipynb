```python
"""Q2. Check for missing values in each column. Identify which columns have null values (especially Last Present Date with "Never") and report the count. """
#-----------------------------Write Solution Code-----------------------------
```

```python
"""Q3. Rename the columns to snake_case (e.g., Roll No → roll_no, Student Name → student_name, Attendance % → attendance_pct). Display the updated column list."""
#-----------------------------Write Solution Code-----------------------------
```

```python
# Rename columns by converting to lowercase, replacing spaces and special characters with underscores
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace('%', 'pct', regex=False)
    .str.replace(r'[^a-z0-9_]', '_', regex=True)
    .str.replace(r'_+', '_', regex=True)
    .str.strip('_')
)

# Display the updated column list
print("Updated columns:")
print(df.columns.tolist())
```

```python
"""Q4. Convert Attendance % from string format (e.g., "64%") to a numeric float column (0.64 or 64.0). Also convert Last Present Date to datetime, handling "Never" as NaT."""
#-----------------------------Write Solution Code-----------------------------
```

```python
# 1. Convert attendance_pct to numeric float
if df['attendance_pct'].dtype == 'object':
    df['attendance_pct'] = (
        df['attendance_pct']
        .astype(str)
        .str.rstrip('%')
        .astype(float)
    )

# 2. Convert last_present_date to datetime, treating 'Never' (and invalid entries) as NaT
df['last_present_date'] = pd.to_datetime(
    df['last_present_date'].replace('Never', None), 
    errors='coerce'
)

# Display data types and sample rows to verify
print(df[['attendance_pct', 'last_present_date']].dtypes)
df[['attendance_pct', 'last_present_date']].head()
```

### Section 3: Grouping & Aggregation 

```python
"""Q9. Group the data by Status and compute: 
a) Count of students 
b) Average attendance %  
c) Maximum consecutive absences"""
#-----------------------------Write Solution Code-----------------------------
```

```python
# Group by 'status' and aggregate count, mean attendance_pct, and max consecutive_absences
# Convert percentage strings (for example, "64%") to numbers before calculating the mean
df['attendance_pct'] = pd.to_numeric(
    df['attendance_pct'].astype(str).str.rstrip('%'),
    errors='coerce'
)

status_summary = df.groupby('status').agg(
    student_count=('student_name', 'count'),
    avg_attendance_pct=('attendance_pct', 'mean'),
    max_consecutive_absences=('consecutive_absences', 'max')
).reset_index()

# Display the summary table
display(status_summary)
```

```python
"""Q10. Create a new column Attendance Category:
≥ 90% → "Excellent"
75-89% → "Good"
60-74% → "Average"
< 60% → "Poor"
Then use groupby() to count students in each category."""
#-----------------------------Write Solution Code-----------------------------
```

```python
import numpy as np

# 1. Define conditions and choices for the category assignment
conditions = [
    df['attendance_pct'] >= 90,
    (df['attendance_pct'] >= 75) & (df['attendance_pct'] < 90),
    (df['attendance_pct'] >= 60) & (df['attendance_pct'] < 75),
    df['attendance_pct'] < 60
]

choices = ['Excellent', 'Good', 'Average', 'Poor']

# 2. Create the 'attendance_category' column using np.select
df['attendance_category'] = np.select(conditions, choices, default='Unknown')

# 3. Group by 'attendance_category' and count the number of students
category_counts = (
    df.groupby('attendance_category')
    .size()
    .reset_index(name='student_count')
)

# Display the category counts
display(category_counts)
```

```python
"""Q12. Group by Status and use .agg() to find:
a)min, max, and mean of Present
b)sum of Absent"""
#-----------------------------Write Solution Code-----------------------------
```

```python
# Select the appropriate present/absent columns dynamically based on existing dataframe columns
present_col = 'total_present' if 'total_present' in df.columns else 'present'
absent_col = 'total_absent' if 'total_absent' in df.columns else 'absent'

# Group by status and calculate requested aggregations
status_agg = df.groupby('status').agg(
    min_present=(present_col, 'min'),
    max_present=(present_col, 'max'),
    mean_present=(present_col, 'mean'),
    sum_absent=(absent_col, 'sum')
).reset_index()

# Display results
display(status_agg)
```

### Section 4: Feature Engineering & Transformation

```python
"""Q13. Create a new column Risk Score using the formula:
Risk Score = (Absent * 0.4) + (Consecutive Absences * 0.6)
Display the top 10 students with the highest risk score."""
#-----------------------------Write Solution Code-----------------------------
```

```python
# Identify column names for absent and consecutive absences dynamically
absent_col = 'total_absent' if 'total_absent' in df.columns else 'absent'
consecutive_col = 'consecutive_absences' if 'consecutive_absences' in df.columns else 'consecutive_absent'

# Calculate Risk Score: (Absent * 0.4) + (Consecutive Absences * 0.6)
df['risk_score'] = (df[absent_col] * 0.4) + (df[consecutive_col] * 0.6)

# Display top 10 students with the highest risk score
top_10_risk = (
    df[['roll_no', 'student_name', absent_col, consecutive_col, 'risk_score']]
    .sort_values(by='risk_score', ascending=False)
    .head(10)
)

display(top_10_risk)
```

```python
"""Q14. Add a column Classes Needed (Numeric) by extracting the numeric value from the Classes Needed (75%) column (e.g., "5 more" → 5, "✓ Met" → 0)."""
#-----------------------------Write Solution Code-----------------------------
```

```python
# Identify column name dynamically
col_75 = [c for c in df.columns if '75' in c][0]

# Extract numeric value or set to 0 for 'Met'
df['classes_needed_numeric'] = (
    df[col_75]
    .astype(str)
    .str.extract(r'(\d+)')
    .fillna(0)
    .astype(int)
)

# Display sample output
display(df[['student_name', col_75, 'classes_needed_numeric']].head(10))
```

```python
"""Q15. Using apply() with a lambda function, create a column Attendance Flag that returns:
a)"Defaulter" if attendance < 75%
b)"Safe" otherwise """
#-----------------------------Write Solution Code-----------------------------
```

```python
# Create Attendance Flag using apply with a lambda function
df['attendance_flag'] = df['attendance_pct'].apply(
    lambda x: "Defaulter" if x < 75 else "Safe"
)

# Display student name, attendance %, and the generated flag
display(df[['student_name', 'attendance_pct', 'attendance_flag']].head(10))
```

```python
"""Q16. Use pd.cut() to bin students into attendance ranges: [0-50, 50-75, 75-90, 90-100] and label them as ["Critical", "Warning", "Good", "Excellent"]."""
#-----------------------------Write Solution Code-----------------------------
```

```python
# Bin attendance into defined ranges using pd.cut
bins = [0, 50, 75, 90, 100]
labels = ["Critical", "Warning", "Good", "Excellent"]

df['attendance_bin'] = pd.cut(
    df['attendance_pct'], 
    bins=bins, 
    labels=labels, 
    include_lowest=True
)

# Display student name, attendance %, and assigned bin
display(df[['student_name', 'attendance_pct', 'attendance_bin']].head(10))
```

### Section 5: Advanced Analysis 

```python
"""Q17. Using pivot_table(), create a summary showing the count of students by Status and Attendance Category (created in Q10)."""
#-----------------------------Write Solution Code-----------------------------
```

```python
# Create a pivot table showing the count of students by Status and Attendance Category
pivot_summary = df.pivot_table(
    index='status', 
    columns='attendance_category', 
    values='student_name', 
    aggfunc='count', 
    fill_value=0
)

# Display the pivot table
display(pivot_summary)
```

```python
"""Q18. Identify duplicate student names (e.g., "ANANYA", "DISHA") and display all rows where the base name repeats. Use string manipulation to extract the base name."""
#-----------------------------Write Solution Code-----------------------------
```

```python
# 1. Extract base name by stripping trailing numbers or extra whitespace
# Handles patterns like "ANANYA 1", "ANANYA 2", "DISHA 1", etc.
df['base_name'] = df['student_name'].str.replace(r'\s*\d+$', '', regex=True).str.strip()

# 2. Identify all rows where the base name is duplicated
duplicate_mask = df.duplicated(subset=['base_name'], keep=False)
duplicate_students = df[duplicate_mask].sort_values(by='base_name')

# Display the duplicate student rows
display(duplicate_students[['roll_no', 'student_name', 'base_name', 'status', 'attendance_pct']])
```

```python
""""Q19. Merge this dataset with a hypothetical Student_Details DataFrame containing Roll No, Branch, and Semester. Perform an inner join and display the merged result."""
#-----------------------------Write Solution Code-----------------------------
```

```python
import pandas as pd

# Create hypothetical Student_Details DataFrame matching roll_no values in df
student_details = pd.DataFrame({
    'roll_no': df['roll_no'].unique(),
    'branch': ['AIDS'] * len(df['roll_no'].unique()),
    'semester': [4] * len(df['roll_no'].unique())
})

# Perform inner join on 'roll_no'
merged_df = pd.merge(df, student_details, on='roll_no', how='inner')

# Display the merged DataFrame
display(merged_df.head(10))
```
